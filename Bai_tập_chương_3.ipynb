### TH 3.2.

Điểm của 8 giám khảo là [7.5, 9.0, 3.0, 8.0, 6.5, 10.0, 7.0, 2.5]. Để giảm ảnh hưởng của giá trị bất thường, bỏ đi $n = 2$ điểm cao nhất và $n = 2$ điểm thấp nhất rồi tính trung bình các điểm còn lại (trung bình cắt đầu–cuối). So sánh với trung bình thông thường; không sửa danh sách gốc.

```python
scores = [7.5, 9.0, 3.0, 8.0, 6.5, 10.0, 7.0, 2.5]
n = 2

# Tính trung bình thông thường
trung_binh_thuong = sum(scores) / len(scores)

# Sắp xếp bản sao của danh sách
sorted_scores = sorted(scores)
# Cắt bỏ n điểm thấp nhất và n điểm cao nhất
trimmed_scores = sorted_scores[n:len(sorted_scores) - n]
# Tính trung bình cắt đầu-cuối
trung_binh_cat = sum(trimmed_scores) / len(trimmed_scores)

print(f"Danh sách gốc: {scores}")
print(f"Danh sách sau khi sắp xếp: {sorted_scores}")
print(f"Danh sách sau khi cắt bỏ 2 đầu: {trimmed_scores}")
print(f"Trung bình thông thường: {trung_binh_thuong}")
print(f"Trung bình cắt đầu-cuối: {trung_binh_cat}")
```

### TH 3.4.

Số giờ ôn tập $x$ và điểm thi $y$ của 5 sinh viên là [(1, 4.0), (2, 5.5), (3, 6.0), (4, 7.5), (5, 8.0)]. Tìm đường hồi quy tuyến tính $y = a + bx$ theo công thức:
$$b = \frac{\sum(x_i - \bar{x})(y_i - \bar{y})}{\sum(x_i - \bar{x})^2}$$
$$a = \bar{y} - b\bar{x}$$
Dự đoán điểm của sinh viên ôn 6 giờ; kiểm tra bằng statistics.linear_regression.

```python
import statistics

data = [(1, 4.0), (2, 5.5), (3, 6.0), (4, 7.5), (5, 8.0)]
x_vals = [pt[0] for pt in data]
y_vals = [pt[1] for pt in data]

# Tính trung bình x và y
x_bar = sum(x_vals) / len(x_vals)
y_bar = sum(y_vals) / len(y_vals)

# Tính tử số và mẫu số cho hệ số b
tu_so = sum((x_i - x_bar) * (y_i - y_bar) for x_i, y_i in data)
mau_so = sum((x_i - x_bar)**2 for x_i in x_vals)

b = tu_so / mau_so
a = y_bar - b * x_bar

# Dự đoán cho x = 6
x_pred = 6
y_pred = a + b * x_pred

print("--- Tự tính toán theo công thức ---")
print(f"Hệ số b: {b}")
print(f"Hệ số a: {a}")
print(f"Đường hồi quy: y = {a} + {b}x")
print(f"Dự đoán điểm thi cho {x_pred} giờ ôn tập: {y_pred}")

# Kiểm tra bằng thư viện statistics
print("\n--- Kiểm tra bằng thư viện statistics ---")
reg_slope, reg_intercept = statistics.linear_regression(x_vals, y_vals)
print(f"Hệ số b (slope): {reg_slope}")
print(f"Hệ số a (intercept): {reg_intercept}")
print(f"Dự đoán bằng thư viện: {reg_intercept + reg_slope * x_pred}")
```

### TH 3.6.

Cho danh sách các cặp (tên, điểm): [("An", 9.0), ("Binh", 7.5), ("Chi", 8.6), ("Dung", 5.0), ("Em", 6.0), ("Phuc", 3.5), ("Giang", 7.0)]. Quy đổi sang điểm chữ ($\ge 8.5$: A; $\ge 7$: B; $\ge 5.5$: C; $\ge 4$: D; còn lại F) rồi gom tên sinh viên theo điểm chữ bằng từ điển (dùng defaultdict(list)). In các nhóm theo thứ tự chữ cái.

```python
from collections import defaultdict

students = [
    ("An", 9.0), ("Binh", 7.5), ("Chi", 8.6),
    ("Dung", 5.0), ("Em", 6.0), ("Phuc", 3.5),
    ("Giang", 7.0)
]

def quy_doi_diem(score):
    if score >= 8.5:
        return 'A'
    elif score >= 7.0:
        return 'B'
    elif score >= 5.5:
        return 'C'
    elif score >= 4.0:
        return 'D'
    else:
        return 'F'

# Gom nhóm bằng defaultdict
nhom_diem = defaultdict(list)
for name, score in students:
    grade = quy_doi_diem(score)
    nhom_diem[grade].append(name)

# Sắp xếp và in ra theo thứ tự chữ cái của điểm chữ
for grade in sorted(nhom_diem.keys()):
    # Sắp xếp tên trong mỗi nhóm để dễ theo dõi
    sorted_names = sorted(nhom_diem[grade])
    print(f"Nhóm {grade}: {sorted_names}")
```

### TH 3.8.

Doanh thu theo sản phẩm của hai cửa hàng:
- Cửa hàng 1: "ao": 120, "quan": 80, "non": 30
- Cửa hàng 2: "ao": 90, "giay": 150, "non": 20

Hợp nhất thành tổng doanh thu theo sản phẩm, in theo thứ tự giảm dần; tìm các sản phẩm bán ở cả hai cửa hàng và các sản phẩm chỉ bán ở một cửa hàng (dùng phép toán tập hợp trên khoá).

```python
ch1 = {"ao": 120, "quan": 80, "non": 30}
ch2 = {"ao": 90, "giay": 150, "non": 20}

# Hợp nhất tổng doanh thu
tong_doanh_thu = ch1.copy()
for sp, dt in ch2.items():
    tong_doanh_thu[sp] = tong_doanh_thu.get(sp, 0) + dt

# Sắp xếp doanh thu giảm dần
sorted_dt = sorted(tong_doanh_thu.items(), key=lambda x: x[1], reverse=True)
print("--- Tổng doanh thu theo sản phẩm (giảm dần) ---")
for sp, dt in sorted_dt:
    print(f"{sp}: {dt}")

# Sử dụng phép toán tập hợp trên khóa
keys1 = ch1.keys()
keys2 = ch2.keys()

ca_hai = keys1 & keys2
chi_mot = keys1 ^ keys2

print("\n--- Phân loại sản phẩm ---")
print(f"Sản phẩm bán ở cả hai cửa hàng: {list(ca_hai)}")
print(f"Sản phẩm chỉ bán ở một cửa hàng: {list(chi_mot)}")
```

### TH 3.10.

Dùng sàng Eratosthenes (cài đặt bằng set) tìm các số nguyên tố không vượt quá $1000$: in số lượng, tổng, và số cặp số nguyên tố sinh đôi $(p, p+2)$ cùng $\le 1000$. Kiểm tra: có $168$ số nguyên tố, tổng bằng $76127$.

```python
def sang_eratosthenes(limit=1000):
    # Bắt đầu với tất cả các số từ 2 đến limit dưới dạng một set
    s_nt = set(range(2, limit + 1))

    # Sàng lọc bội số bắt đầu từ p = 2
    for p in range(2, int(limit**0.5) + 1):
        if p in s_nt:
            # Loại bỏ bội số lớn hơn hoặc bằng p^2 của p
            boi_so = set(range(p*p, limit + 1, p))
            s_nt -= boi_so
    return s_nt

# Tìm tập hợp các số nguyên tố
primes_set = sang_eratosthenes(1000)
primes_sorted = sorted(list(primes_set))

# Tính số lượng và tổng
so_luong = len(primes_set)
tong_nt = sum(primes_set)

# Đếm cặp số nguyên tố sinh đôi (p, p+2) trong giới hạn <= 1000
cap_sinh_doi = []
for p in primes_sorted:
    if p + 2 <= 1000 and (p + 2) in primes_set:
        cap_sinh_doi.append((p, p + 2))

print(f"Số lượng số nguyên tố <= 1000: {so_luong} (Kỳ vọng: 168)")
print(f"Tổng các số nguyên tố <= 1000: {tong_nt} (Kỳ vọng: 76127)")
print(f"Số lượng cặp số nguyên tố sinh đôi: {len(cap_sinh_doi)}")
print(f"Một số cặp nguyên tố sinh đôi tiêu biểu: {cap_sinh_doi[:10]}...")

# Assert kiểm chứng kết quả lý thuyết
assert so_luong == 168
assert tong_nt == 76127
```
