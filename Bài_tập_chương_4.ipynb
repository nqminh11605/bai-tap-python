### TH 4.1: Viết lớp HinhCau có bán kính ban_kinh, các phương thức dien_tich(), the_tich() và __ str __. Hàm tạo phải báo ValueError nếu bán kính không dương.

```python
import math

class HinhCau:
    def __init__(self, ban_kinh):
        if ban_kinh <= 0:
            raise ValueError("Ban kinh phai la so duong")
        self.ban_kinh = ban_kinh

    def dien_tich(self):
        return 4 * math.pi * (self.ban_kinh ** 2)

    def the_tich(self):
        return (4 / 3) * math.pi * (self.ban_kinh ** 3)

    def __str__(self):
        return f"Hinh cau (r={self.ban_kinh})"

hc = HinhCau(2)
print(hc)
print("Dien tich:", hc.dien_tich())
print("The tich:", hc.the_tich())

try:
    hc_loi = HinhCau(-1)
except ValueError as e:
    print("Loi phat hien dung:", e)
```

### TH 4.3: Định nghĩa ngoại lệ SoDuKhongDu và lớp TaiKhoan với các phương thức nạp, rút tiền cùng lịch sử giao dịch.

```python
class SoDuKhongDu(Exception):
    pass

class TaiKhoan:
    def __init__(self, chu_tai_khoan, so_du_ban_dau=0):
        if so_du_ban_dau < 0:
            raise ValueError("So du ban dau khong duoc am")
        self.chu_tai_khoan = chu_tai_khoan
        self.so_du = so_du_ban_dau
        self.lich_su = [f"Khoi tao: {so_du_ban_dau}"]

    def nap(self, tien):
        if tien <= 0:
            raise ValueError("So tien nap phai lon hon 0")
        self.so_du += tien
        self.lich_su.append(f"Nap: {tien}")

    def rut(self, tien):
        if tien <= 0:
            raise ValueError("So tien rut phai lon hon 0")
        if tien > self.so_du:
            raise SoDuKhongDu("So du khong du")
        self.so_du -= tien
        self.lich_su.append(f"Rut: {tien}")

tk = TaiKhoan("Nguyen Van A", 1000000)
tk.nap(500000)
tk.rut(200000)
print("So du hien tai:", tk.so_du)

try:
    tk.rut(2000000)
except SoDuKhongDu as e:
    print("Phat hien loi rut qua so du:", e)

print("Lich su:", tk.lich_su)
```

### TH 4.5: Thiết kế lớp cha HinhHoc và 3 lớp con: HinhTron, HinhChuNhat, HinhTamGiac. Sắp xếp danh sách hình theo diện tích tăng dần và tính tổng diện tích.

```python
import math

class HinhHoc:
    def dien_tich(self):
        raise NotImplementedError()

class HinhTron(HinhHoc):
    def __init__(self, r):
        self.r = r
    def dien_tich(self):
        return math.pi * (self.r ** 2)
    def __str__(self):
        return f"Hinh tron r={self.r}"

class HinhChuNhat(HinhHoc):
    def __init__(self, a, b):
        self.a = a
        self.b = b
    def dien_tich(self):
        return self.a * self.b
    def __str__(self):
        return f"Hinh chu nhat {self.a}x{self.b}"

class HinhTamGiac(HinhHoc):
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c
    def dien_tich(self):
        p = (self.a + self.b + self.c) / 2
        return math.sqrt(p * (p - self.a) * (p - self.b) * (p - self.c))
    def __str__(self):
        return f"Hinh tam giac {self.a},{self.b},{self.c}"

danh_sach = [HinhTron(1), HinhChuNhat(3, 4), HinhTamGiac(3, 4, 5)]

danh_sach_sap_xep = sorted(danh_sach, key=lambda h: h.dien_tich())

tong_dt = 0
for h in danh_sach_sap_xep:
    dt = h.dien_tich()
    tong_dt += dt
    print(h, "- Dien tich:", dt)

print("Tong dien tich:", tong_dt)
```

### TH 4.7: Đọc tệp data/goldprice.csv (nếu chưa có thư mục và tệp, ta sẽ giả lập tạo nhanh dữ liệu mẫu để chạy minh họa), tính giá trung bình các năm đủ 12 tháng, tìm năm tăng mạnh nhất và giảm mạnh nhất.

```python
import csv

gia_theo_nam = {}

with open("data/goldprice.csv", "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    next(reader)
    for row in reader:
        date, price = row[0], float(row[1])
        year = int(date.split("-")[0])
        if year not in gia_theo_nam:
            gia_theo_nam[year] = []
        gia_theo_nam[year].append(price)

trung_binh_nam = {}
for year, prices in gia_theo_nam.items():
    if len(prices) == 12:
        trung_binh_nam[year] = sum(prices) / 12

tang_truong = {}
danh_sach_nam = sorted(trung_binh_nam.keys())
for i in range(1, len(danh_sach_nam)):
    nam_nay = danh_sach_nam[i]
    nam_truoc = danh_sach_nam[i-1]
    pct = ((trung_binh_nam[nam_nay] - trung_binh_nam[nam_truoc]) / trung_binh_nam[nam_truoc]) * 100
    tang_truong[nam_nay] = pct

sap_xep_tang = sorted(tang_truong.items(), key=lambda item: item[1], reverse=True)

print("Top 3 nam tang manh nhat:")
for year, val in sap_xep_tang[:3]:
    print(f"Nam {year}: Tang {val:.2f}%")

print("\nNam giam manh nhat:")
print(f"Nam {sap_xep_tang[-1][0]}: Giam {sap_xep_tang[-1][1]:.2f}%")
```

### TH 4.9: Tạo tệp luong.csv theo định dạng phân tách gạch chéo sang dòng. Sử dụng ngoại lệ DongKhongHopLe để phát hiện và báo lỗi dòng không hợp lệ.

```python
import csv

class DongKhongHopLe(Exception):
    pass

valid_rows = []
invalid_rows = []
total_salary = 0

with open("luong.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for idx, row in enumerate(reader, start=2):
        try:
            nam_kn = float(row["nam_kn"])
            luong = float(row["luong"])
            valid_rows.append((nam_kn, luong))
            total_salary += luong
        except (ValueError, TypeError):
            invalid_rows.append(idx)

print("Dong hop le:", valid_rows)
print("Dong loi:", invalid_rows)
if valid_rows:
    print("Luong trung binh:", total_salary / len(valid_rows))
```
