### Bài 2.4.
**(Dự đoán kết quả)** Hàm đệ quy dưới đây trả về cặp (giá trị, số lời gọi). Dự đoán kết quả của `g(2468)`, `g(0)` và `g(1999)`. Hàm `g` thực chất tính đại lượng gì của `n`?

Hàm `g(n)` thực chất tính tổng các chữ số của số nguyên dương $n$ và trả về cùng với số chữ số cộng thêm 1 (chính là tổng số bước gọi đệ quy bao gồm cả bước neo `n == 0`).

```python
def g(n):
    if n == 0:
        return 0, 1
    v, c = g(n // 10)
    return n % 10 + v, c + 1

print(g(2468), g(0), g(1999))
```

### Bài 2.7.
**(Đọc lỗi và sửa mã)** Chương trình tính `100/so_ban` phải xử lý được cả đầu vào không phải số (abc) lẫn 0. Chạy mã với hai đầu vào đó, giải thích các lỗi còn sót (có hai lỗi khác nhau) và sửa.

**Giải thích lỗi bài:**
Mã gốc chưa xử lý triệt để hai trường hợp lỗi sau:
1. **Lỗi nhập chữ (`abc`)**: Hàm `int(input(...))` ném ra lỗi `ValueError` khi người dùng nhập chuỗi ký tự. Khối ngoại lệ `except ValueError` đã bắt được lỗi này, nhưng biến `tb` lúc này chưa được định nghĩa khiến câu lệnh `print(f"Kết quả: {tb}")` bên ngoài ném ra lỗi tiếp theo là `NameError: name 'tb' is not defined`.
2. **Lỗi chia cho 0 (`0`)**: Khi nhập số `0`, chương trình ném ra lỗi `ZeroDivisionError` vì ta không thể chia cho 0. Khối mã gốc chưa có bộ bắt lỗi này khiến chương trình bị crash trực tiếp khi gặp đầu vào bằng `0`.

```python
try:
    so_ban = int(input("Số bạn: "))
    tb = 100 / so_ban
except ValueError:
    print("Hãy nhập số nguyên!")
except ZeroDivisionError:
    print("Số bạn phải khác 0!")
    tb = None

if 'tb' in locals() and tb is not None:
    print(f"Kết quả: {tb}")
```

### Bài 2.10.
**(Bẫy logic)** Chương trình dưới đây muốn in "x là 1 hoặc 2" khi $x \in \{1, 2\}$, nhưng với $x = 5$ vẫn in như vậy. Giải thích bằng quy tắc ưu tiên toán tử và giá trị "đúng/sai" của số, rồi viết hai cách sửa đúng.

**Giải thích:**
- Quy tắc ưu tiên toán tử: Biểu thức so sánh `==` có độ ưu tiên cao hơn toán tử logic `or`. Do đó, biểu thức `x == 1 or 2` được Python hiểu và phân tích thành: `(x == 1) or (2)`.
- Giá trị chân trị (truthy): Trong Python, các số khác `0` (như số `2`) đều được đánh giá là `True` về mặt logic. Do đó, biểu thức luôn tương đương với `(x == 1) or True` -> luôn trả về kết quả `True` bất kể giá trị của `x`.

```python
x = 5
# Cách sửa 1:
if x == 1 or x == 2:
    print("x là 1 hoặc 2")
else:
    print("x khác 1 và 2")
```

```python
x = 5
# Cách sửa 2:
if x in {1, 2}:
    print("x là 1 hoặc 2")
else:
    print("x khác 1 và 2")
```

### Bài 2.11.
**(Bẫy logic)** Hàm `tao_nhom` cho mỗi lần gọi một nhóm mới chứa tên truyền vào. Dự đoán rồi chạy; vì sao hai nhóm lại giống nhau? Sửa hàm để `a` và `b` là hai danh sách độc lập.

**Nguyên nhân**: các giá trị mặc định của tham số (như `thanh_vien=[]`) chỉ được khởi tạo một lần duy nhất khi định nghĩa hàm chứ không phải mỗi lần hàm được gọi. Do đó, tất cả các lần gọi hàm không truyền đối số `thanh_vien` đều sẽ dùng chung và thao tác trên cùng một đối tượng danh sách (list) duy nhất trong bộ nhớ.

```python
def tao_nhom(ten, thanh_vien=None):
    if thanh_vien is None:
        thanh_vien = []
    thanh_vien.append(ten)
    return thanh_vien

a = tao_nhom("An")
b = tao_nhom("Bình")
print(a, b, a is b)
```
