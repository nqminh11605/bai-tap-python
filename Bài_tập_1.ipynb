
### Bài 1.1 (Dự đoán kết quả)
Với $x = 23$ và $y = 6$, hãy tính tay giá trị các biểu thức `x / y`, `x // y`, `x % y`, `x ** 2`, `-x // y`, `-x % y`, `2 ** 3 ** 2`, `(2 ** 3) ** 2`; sau đó kiểm tra lại bằng máy


```python
x = 23
y = 6
print("x / y =", x / y)
print("x // y =", x // y)
print("x % y =", x % y)
print("x ** 2 =", x ** 2)
print("-x // y =", -x // y)
print("-x % y =", -x % y)
print("2 ** 3 ** 2 =", 2 ** 3 ** 2)
print("(2 ** 3) ** 2 =", (2 ** 3) ** 2)
```

### Bài 1.2 (Dự đoán kết quả)
Cho `s = "KY THUAT LAP TRINH"`. Dự đoán kết quả của `len(s)`, `s[0]`, `s[-1]`, `s[3:8]`, `s[:2]`, `s[::-1][:4]`, `s[::3]` và `s[30:]`. Kiểm tra lại bằng máy.

```python
s = "KY THUAT LAP TRINH"
print("len(s) =", len(s))
print("s[0] =", s[0])
print("s[-1] =", s[-1])
print("s[3:8] =", s[3:8])
print("s[:2] =", s[:2])
print("s[::-1][:4] =", s[::-1][:4])
print("s[::3] =", s[::3])
print("s[30:] =", s[30:])
```

### Bài 1.3 (Dự đoán kết quả)
Dự đoán đầu ra của bốn dòng dưới đây rồi chạy để kiểm tra. Chú ý các giá trị làm tròn và phép ép kiểu; giải thích vì sao `round(2.5)` và `round(3.5)` không cho kết quả “tăng một đơn vị” như nhau.

```python
print(round(2.5), round(3.5), round(0.5), round(1.2345, 2))
print(int(-3.7), -3.7 // 1, int(3.99))
print(int("12") + int("3"), "12" + "3")
print("ab" * 3 + "c", 7 / 2 * 2, 10 - 2 ** 2 * 3)

# Giải thích: Python sử dụng phương pháp "làm tròn về số chẵn gần nhất" (Bankers' rounding) cho các số kết thúc bằng .5.
# Do đó, round(2.5) sẽ làm tròn về 2 (số chẵn gần nhất), còn round(3.5) sẽ làm tròn về 4 (số chẵn gần nhất).
```

### Bài 1.4 (Dự đoán kết quả)
Dự đoán chính xác những gì chương trình dưới đây in ra (chú ý khoảng trắng, dấu xuống dòng và định dạng số), rồi chạy kiểm tra.

```python
a, b = 7, 2
print(a, b, sep="")
print(a / b, a // b, sep=" | ")
print(f"{a}+{b}={a + b}", end="; ")
print(f"{a / b:.1f}")
print("a" "b", 3 * "ab")
print(f"{1234.5678:,.2f}", f"{5:03d}", f"{0.256:.1%}")
```

### Bài 1.5 (Đọc lỗi và sửa mã)
Đoạn mã dưới đây tính chỉ số khối cơ thể $BMI = cân\ nặng\ (kg) / chiều\ cao\ (m)^2$. Chạy với đầu vào 60 và 1.7, giải thích ít nhất 2 lỗi sai và sửa lại để ra kết quả BMI = 20.8.

```python
# Lỗi 1: Hàm input() trả về dữ liệu kiểu chuỗi (string), cần ép kiểu sang float để tính toán.
# Lỗi 2: Trong Python, toán tử lũy thừa là ** chứ không phải ^ (toán tử ^ là phép XOR bit).

can_nang = 60.0
chieu_cao = 1.7
bmi = can_nang / (chieu_cao ** 2)
print(f"BMI = {bmi:.1f}")
```
